## 13. Limitações e riscos de interpretação

1. **Atribuição heterogênea.** ~70,7% da receita está em linhas com exceção manual, e o mix de atribuição muda em janeiro e julho/2026. Comparações temporais são afetadas.
2. **Investimento parcial.** 33,47% das linhas das plataformas analisadas não têm investimento (tratado como 0), e o investimento das 5 plataformas excluídas não entra no ROAS. O ROAS pode estar superestimado. Não foi medido quanto da receita está em linhas sem investimento.
3. **Concentração extrema.** Uma campanha responde por ~72% da receita e três grupos de marca por ~92%. Estatísticas médias são pouco representativas.
4. **Atribuição só em Conversão.** Não é possível avaliar a contribuição de Awareness e Consideração com estes dados.
5. **Preenchimento de nulos com zero.** Pode confundir "sem receita" com "dado ausente".
6. **Relações descritivas.** Correlações não indicam causalidade e `orders`/`visits` são mecanicamente ligadas à receita.
7. **Escopo de plataformas.** Awin, Kayak, Skyscanner, DV360 e Vivo ficaram fora da análise de receita.
8. **Código de exploração.** No notebook de análise, a variável `nulos_receita` é reatribuída (na seção 2.4) com os nulos de `investimento`, e uma célula posterior de percentual por grupo usa essa versão. Os resultados de receita desta análise não dependem disso, mas vale corrigir os nomes.
