```python
total_linhas_plataforma = df.groupby('platform').size()

nulos_receita = df[df["receita"].isnull()]
nulos_por_plataforma = nulos_receita.groupby('platform').size()

pct_nulo_plataforma = nulos_por_plataforma / total_linhas_plataforma * 100
pct_nulo_plataforma.sort_values(ascending=False)
```

Algumas plataformas (Awin, Kayak, Skyscanner, e em menor grau DV360, Vivo e RTB) têm taxas de
nulo muito altas (a partir de 59%, chegando a 100%). Isso não parece ser falha aleatória de coleta — é mais
provável que sejam canais onde a receita simplesmente não é atribuída diretamente na origem dos dados.

Já Google Ads, UOL, Taboola, Meta Ads, Bing Ads e Criteo têm taxas de nulo bem mais baixas (abaixo de 40%),
indicando que a atribuição de receita funciona de forma mais direta nesses canais.
