report_date: data a que se referem os dados daquela linha do relatório.

campaign_key: identificador único da campanha, combinando plataforma, conta, canal e um id ou nome, no formato plataforma|conta|canal|identificador.

campaign_name: nome legível da campanha, como aparece na plataforma de mídia.

platform: plataforma de mídia onde a campanha foi veiculada, como Google Ads, Meta Ads, Bing Ads, DV360, Criteo, Taboola, Awin, RTB, Kayak, Skyscanner, UOL ou Vivo.

campaign_group: agrupamento ou segmento ao qual a campanha pertence, geralmente indicando o público-alvo ou a estratégia usada, como marca, concorrentes ou genérico.

funnel_stage: etapa do funil de marketing à qual a campanha corresponde, podendo ser Awareness, Consideração, Conversão, ou não identificado quando não foi possível classificar.

campaign_key_type: indica se a campanha foi identificada usando um id numérico ou usando o nome da campanha, já que nem toda plataforma fornece um id estável.

receita: valor total de receita gerada e atribuída àquela campanha no período.

orders: número de pedidos ou conversões atribuídos à campanha.

visits: número de visitas ao site geradas pela campanha.

investimento: valor total investido em mídia naquela campanha.

impressions: número de vezes que os anúncios da campanha foram exibidos.

clicks: número de cliques recebidos pelos anúncios da campanha.

roas: retorno sobre o investimento em mídia, calculado como receita dividida por investimento.

receita_safe_ad: parte da receita atribuída com alta confiança, porque foi possível identificar com segurança o anúncio responsável pela venda.

receita_excecao_manual: parte da receita atribuída por meio de uma regra ou ajuste manual definido à parte.

receita_pmax_sem_anuncio: parte da receita vinda de campanhas Performance Max do Google, que não permitem identificar o anúncio individual responsável.

receita_por_nome_campanha: parte da receita atribuída por meio da correspondência do nome da campanha, e não de um id.

receita_outro: parte da receita atribuída por algum outro método não coberto pelas categorias anteriores.

receita_anuncio_ambiguo: parte da receita vinda de anúncios cuja atribuição é ambígua, ou seja, não é possível garantir com certeza de qual campanha ela veio.

receita_nao_casado: parte da receita que não pôde ser associada a nenhuma campanha com segurança.

investimento_anuncio: parte do investimento que foi atribuída no nível de anúncio individual.

investimento_campanha: parte do investimento que foi atribuída apenas no nível de campanha, sem detalhe de anúncio.

pct_receita_assertiva: percentual da receita total que foi atribuído com alta confiança, considerando os métodos mais seguros de atribuição.

pct_receita_anuncio_ambiguo: percentual da receita total que veio de atribuição ambígua.

pct_receita_assertiva_alta: percentual da receita total que foi atribuído com o nível mais alto de confiança possível.

pct_investimento_anuncio: percentual do investimento total que foi atribuído no nível de anúncio, em vez de apenas no nível de campanha.

motivos_receita: lista dos métodos ou motivos que foram usados para compor a receita daquela linha, podendo combinar mais de um.

motivos_investimento: indica se o investimento daquela linha veio do nível de anúncio, do nível de campanha, de ambos, ou está vazio.
