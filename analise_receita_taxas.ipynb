# 3. Tratamento

## 3.1 Datas

```python
df['report_date'] = df['report_date'].str.strip()
df["report_date"] = pd.to_datetime(df["report_date"], format="%Y-%m-%d")
```

```python
df.isnull().sum()
```

```python
df_receita.isnull().sum() 
```

```python
(df_receita.isnull().sum() / len(df_receita)) * 100
```

```python
df_receita_sem_pct = df_receita.drop(columns=['pct_receita_assertiva', 'pct_receita_anuncio_ambiguo', 'pct_receita_assertiva_alta', 'pct_investimento_anuncio'])
```

```python
(df_receita_sem_pct.isnull().sum() / len(df_receita)) * 100
```

Como o objetivo da análise é analisar oque afeta a receita vão ser desconsideradas as variáveis de porcentagem.

```python
df_receita_sem_pct.head()
```

```python
numericas_zero = [
    "receita", "orders", "visits", "investimento", "impressions", "clicks",
    "receita_safe_ad", "receita_excecao_manual", "receita_pmax_sem_anuncio",
    "receita_por_nome_campanha", "receita_outro", "receita_anuncio_ambiguo", "receita_nao_casado",
    "investimento_anuncio", "investimento_campanha",
]

df_receita_sem_pct_fill_num = df_receita_sem_pct.copy()

for col in numericas_zero:
    df_receita_sem_pct_fill_num[col] = df_receita_sem_pct_fill_num[col].fillna(0)
```

### Preenchimento dos nulos categóricos

```python
df_receita_sem_pct_fill_num["motivos_receita"] = df_receita_sem_pct_fill_num["motivos_receita"].fillna("sem_receita")
df_receita_sem_pct_fill_num["motivos_investimento"] = df_receita_sem_pct_fill_num["motivos_investimento"].fillna("sem_investimento")
```

```python
(df_receita_sem_pct_fill_num.isnull().sum() / len(df_receita)) * 100
```

Roas recalculadas

```python

df_receita_sem_pct_fill_num["roas"] = df_receita_sem_pct_fill_num["receita"] / df_receita_sem_pct_fill_num["investimento"].replace(0, np.nan)
df_receita_sem_pct_fill_num["roas"] = df_receita_sem_pct_fill_num["roas"].fillna(0)
tratada = df_receita_sem_pct_fill_num
```

### Colunas de mês e ano

```python
tratada["report_date"] = pd.to_datetime(tratada["report_date"].str.strip(), format="%Y-%m-%d")

tratada["mes"] = tratada["report_date"].dt.month
tratada["ano"] = tratada["report_date"].dt.year
tratada["ano_mes"] = tratada["report_date"].dt.to_period("M").astype(str)
```

```python
nao_ident = tratada[tratada['funnel_stage'] == 'NÃO IDENTIFICADO'] 
tratada = tratada[tratada['funnel_stage'] != 'NÃO IDENTIFICADO']    
```

Como haviam apenas 61 linhas com 'funnel_stage' = 'NÃO IDENTIFICADO', e elas representavam uma porcentagem muito pequena. Foram apagadas.

**ROI líquido muito bom**

ROI bruto (receita / investimento) = **13,58** — ou seja, para cada R$ 1 investido, a base retornou R$ 13,58 em receita atribuída. O ROI líquido (retornando o investimento já descontado) fica em **12,58**, o que significa que a receita superou o investimento em quase 13 vezes o valor gasto (R$ 756,5 milhões de receita contra R$ 55,7 milhões de investimento).

Esse número precisa de duas ressalvas antes de ser lido como "a operação é extremamente lucrativa":

1. É **ROI de mídia**, não de negócio: não inclui custo do produto/serviço vendido, impostos, comissões de plataforma ou qualquer outro custo operacional — só compara receita atribuída à mídia contra o investimento em mídia.
2. É puxado majoritariamente por Google Ads e Bing Ads (que concentram ~92% do investimento e ~99,5% da receita nesta base) — o número agregado esconde que outras plataformas (Meta Ads, RTB, UOL) têm ROAS abaixo de 1, ou seja, dando prejuízo de mídia isoladamente.

## 7.4 Teste de hipótese: a diferença de ROAS entre Bing e Google é real ou é ruído?

**A lógica do teste:** quando comparamos duas médias (ex: ROAS médio do Bing vs. do Google), a diferença pode ser um padrão real ou pode ser só variação ao acaso, principalmente se os dados tiverem muitos valores extremos (como o nosso ROAS por linha, que vai de 0 a 454+). O teste t de duas amostras responde: "essa diferença de média é grande o suficiente, dado o quanto os dados variam, pra eu acreditar que não é coincidência?"

**Passo a passo:**
1. Separar o `roas` (por linha) de cada plataforma em duas listas.
2. Rodar o teste t, que compara as médias levando em conta o tamanho da amostra e a variância de cada grupo.
3. Olhar o `p-valor`: por convenção, se for menor que 0,05, a diferença é considerada estatisticamente significativa (improvável de ser só acaso).

```python
from scipy import stats

roas_bing = tratada[tratada['platform'] == 'Bing Ads']['roas']
roas_google = tratada[tratada['platform'] == 'Google Ads']['roas']

t, p = stats.ttest_ind(roas_bing, roas_google, equal_var=False)

print(f"Bing Ads   -> n={len(roas_bing)}, média={roas_bing.mean():.2f}, desvio padrão={roas_bing.std():.2f}")
print(f"Google Ads -> n={len(roas_google)}, média={roas_google.mean():.2f}, desvio padrão={roas_google.std():.2f}")
print(f"\nEstatística t = {t:.3f}   p-valor = {p:.4f}")
```

**Resultado: p = 0,31 — não é estatisticamente significativo.** Ou seja, apesar do ROAS médio parecer bem diferente (Bing R$ 6,98 vs. Google R$ 39,80, por linha), o teste diz que **não dá pra afirmar com confiança que essa diferença é real**, e não coincidência.

O motivo aparece nos números: o desvio padrão do ROAS por linha é gigantesco (87,93 no Bing, 2.972,63 no Google) — muito maior que a própria média. Isso é o mesmo problema do ROAS de 454 que já discutimos: poucas linhas com valores extremos bagunçam a média e a variância, tornando qualquer comparação simples de médias pouco confiável nessa métrica calculada por linha.

**Isso não invalida o ROAS agregado da seção 4.2** (receita total / investimento total por plataforma, que deu 11,52 vs. 15,11) — aquele é uma conta diferente, mais estável, porque soma tudo antes de dividir, em vez de tirar média das divisões linha a linha. A lição aqui é: **ROAS por linha é uma métrica ruidosa demais pra comparar estatisticamente**; ROAS agregado (receita total/investimento total) é mais confiável pra esse tipo de comparação.

Repetindo o mesmo teste, mas para a taxa de conversão (que não sofre do mesmo problema de outliers extremos):

```python
sub = tratada[tratada['visits'] > 0].copy()
sub['taxa_conv'] = sub['orders'] / sub['visits']

conv_bing = sub[sub['platform'] == 'Bing Ads']['taxa_conv']
conv_google = sub[sub['platform'] == 'Google Ads']['taxa_conv']

t2, p2 = stats.ttest_ind(conv_bing, conv_google, equal_var=False)
print(f"Estatística t = {t2:.3f}   p-valor = {p2:.6f}")
```

**Resultado: p < 0,0001 — significativo.** Diferente do ROAS por linha, a diferença de taxa de conversão entre Bing e Google **é real**, não coincidência. Isso reforça que a taxa de conversão (seção 7.2) é uma métrica mais confiável pra comparar as duas plataformas do que o ROAS por linha.

## 7.5 Regressão linear: o investimento prediz a receita, mesmo controlando pela plataforma?

**A lógica da regressão:** groupby mostra a média de receita por plataforma, mas não separa "efeito de investir mais" de "efeito de ser uma plataforma melhor". Regressão linear múltipla resolve isso: ela estima o efeito de cada variável **mantendo as outras fixas**.

**Passo a passo:**
1. Transformar `platform` em colunas binárias (0/1) — uma técnica chamada *dummy encoding*, necessária porque regressão só entende número, não texto. Usamos `drop_first=True` pra evitar redundância (uma das plataformas vira a "referência" implícita).
2. Montar a equação: `receita = intercepto + coeficiente₁ × investimento + coeficiente₂ × plataforma_X + ...`
3. Resolver a equação por mínimos quadrados (o método padrão de regressão linear — minimiza a soma dos erros ao quadrado entre previsto e real).
4. Olhar o `p-valor` de cada coeficiente: ele diz se aquela variável, especificamente, tem efeito significativo sobre a receita, controlando as demais.

```python
import numpy as np
from scipy import stats

X = pd.get_dummies(tratada[['investimento', 'platform']], columns=['platform'], drop_first=True)
X = X.astype(float)
X.insert(0, 'intercepto', 1.0)
y = tratada['receita'].astype(float).values
Xm = X.values

# Regressão por mínimos quadrados (OLS)
beta, _, _, _ = np.linalg.lstsq(Xm, y, rcond=None)
n, k = Xm.shape
residuos = y - Xm @ beta
sigma2 = (residuos @ residuos) / (n - k)
erro_padrao = np.sqrt(np.diag(sigma2 * np.linalg.inv(Xm.T @ Xm)))
t_stats = beta / erro_padrao
p_valores = 2 * (1 - stats.t.cdf(np.abs(t_stats), df=n - k))

resultado = pd.DataFrame({'coeficiente': beta, 'erro_padrao': erro_padrao, 't': t_stats, 'p_valor': p_valores}, index=X.columns)
resultado.round(4)
```

**Como ler o resultado:**

- **`investimento`: coeficiente ≈ 16,24, p-valor < 0,0001.** Altamente significativo — cada R$ 1 a mais de investimento está associado, em média, a R$ 16,24 a mais de receita, **mesmo controlando qual é a plataforma**. Isso confirma, com rigor estatístico, que investimento realmente prediz receita — não é só coincidência de que quem investe mais também vende mais por outro motivo.
- **As variáveis de plataforma (Google Ads, Meta Ads, RTB, etc.) não são individualmente significativas** (todos os p-valores > 0,18, bem acima de 0,05). Isso quer dizer que, **depois de considerar o investimento**, não dá pra afirmar que uma plataforma tem uma receita "base" diferente da referência (Bing Ads, usada aqui como grupo de comparação).

**Interpretação com cuidado:** isso não contradiz o ROAS diferente entre plataformas (Google 15,11 vs. Bing 11,52) — o modelo acima assume que a *relação* entre investimento e receita (o "quanto retorna por real investido") é a mesma pra todas as plataformas, e só testa se o "ponto de partida" (intercepto) muda. Pra testar se a **eficiência em si** (a inclinação, não só o ponto de partida) varia por plataforma, seria preciso um modelo com interação (`investimento × platform`) — um passo natural a mais, que pode ficar pra próxima etapa de aprofundamento.


# 8. ROAS por faixa de investimento


Como os valores investidos foram muito dispersos e fica dificil analisar as roas não considerando a proporçao do investimento. Por isso foi feito uma divisão por quartis. Porém devido ao grande número de linhas que não houve investimento, foi necessário criar uma divisão com apenas as linhas sem investimento nenhum.

```python
conv = tratada[tratada['funnel_stage'] == 'Conversão'].copy()

# grupo 0 separado, quartis só nas linhas com gasto
com_gasto = conv['investimento'] > 0
conv['faixa_inv'] = 'Sem investimento'
conv.loc[com_gasto, 'faixa_inv'] = pd.qcut(
    conv.loc[com_gasto, 'investimento'], 4,
    labels=['Q1', 'Q2', 'Q3', 'Q4']
).astype(str)

ordem = ['Sem investimento', 'Q1', 'Q2', 'Q3', 'Q4']

resumo = conv.groupby('faixa_inv').agg(
    linhas=('receita', 'size'),
    inv_min=('investimento', 'min'),
    inv_max=('investimento', 'max'),
    receita=('receita', 'sum'),
    investimento=('investimento', 'sum'),
).reindex(ordem)

# ROAS só onde há investimento
resumo['roas'] = resumo['receita'] / resumo['investimento'].replace(0, np.nan)
resumo['pct_receita'] = resumo['receita'] / resumo['receita'].sum() * 100
resumo.round(2)
```

Sem Investimento - ROAS nulo na linha 'Sem investimento' por conta da falta de investimento para calcular ROAS. Apesar disso, concentra 9,56% da receita total, indicando que uma parcela relevante da receita ocorreu sem investimento registrado.

Q1 - Segunda melhor ROAS, com 12,77, porém representa apenas 0,58% da receita total, devido ao baixo investimento. Isso mostra um retorno proporcionalmente alto, mas sobre um volume pequeno de investimento.

Q2 - Apresenta um ROAS de 4,84, inferior ao Q1, mas ainda significa que, para cada R$ 1 investido, foram gerados aproximadamente R$ 4,84 em receita. Representa 1,04% da receita total, com um investimento maior que o Q1.

Q3 - Possui o menor ROAS entre as faixas com investimento, de 3,45, indicando um retorno proporcionalmente menor. Mesmo com investimento de aproximadamente R$ 4,89 milhões, a faixa representa 2,23% da receita total.

Q4 - Concentra a maior parte dos resultados, representando 86,60% da receita total e R$ 45,48 milhões em investimento. Apresenta ROAS de 14,40, o maior entre todas as faixas com investimento, indicando um alto retorno proporcional e também um grande volume absoluto de receita.
